```python
import pandas as pd

df_loja = pd.read_json("/content/loja.json")
df_pets = pd.read_json("/content/pets.json")
df_produtos = pd.read_json("/content/produtos.json")
df_servicos = pd.read_json("/content/servicos.json")
df_tutores = pd.read_json("/content/tutores.json")


display(df_tutores)
```

```python
#pets + tutores
#loja + produto + serviço

#regra da relação - tem qye ter coisas em comum
#pri,eiro - criar a base "merge"
df_pets_tutores = pd.merge(
 df_pets,
 df_tutores,
 left_on="tutor_id",
 right_on="id"
)

display(df_pets_tutores)
```

```python
import requests # <--- tem essa linha
# esta usando API

#o caminho é chamado de endpoint
endpoint = "https://api.open-meteo.com/v1/forecast"

#param
parametros = {
    "latitude": - 27.7,
    "longitude": -47.55,
    "current":"temperature_2m",
    "timezone":"America/Sao_Paulo"
}

resposta = requests.get(endpoint, params=parametros)


df_dados_climaticos = resposta.json()
display(df_dados_climaticos)
```
