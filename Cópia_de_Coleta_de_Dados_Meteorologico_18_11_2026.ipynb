```python
# https://docs.google.com/spreadsheets/d/1mmogNWapCSNm9U7hKQI41Gopq-_uX0aKCqt8a6oPcF8/edit?usp=sharing

import pandas as pd
import requests
import time

endpoint_planilha = "https://docs.google.com/spreadsheets/d/1mmogNWapCSNm9U7hKQI41Gopq-_uX0aKCqt8a6oPcF8/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
print('---Dados Coletados -----')
dados_planilha = pd.read_csv(endpoint_planilha)

display(dados_planilha)
```

```python
# iterar cada resposta do formulario
# 0 é a posicao ou index, ou indice

for indice, linha in dados_planilha.iterrows():
  nome_local = linha['Nome do local ']
  lat = linha['Latitude']
  lon = linha['Longitude']

  parametros = {
      "latitude":lat ,
      "longitude":lon ,
      "current":'temperature_2m'
  }
  print(f'----{nome_local} ------')

  resposta = requests.get(endpoint_api, params=parametros).json()

  display(resposta)
  print('\n\n')
```
